```python
class Student:
    def __init__(self, name, group, grades, tel_number):
        self.name = name
        self.group = group
        self.grades = grades
        self.__tel_number = tel_number

    def add_grade(self, grade):
        self.grades.append(grade)

    def get_average(self):
        if not self.grades:
            return 0
        return sum(self.grades) / len(self.grades)

    def info(self):
        return (f"Студент: {self.name} | Группа: {self.group} | "
                f"Оценки: {self.grades} | Средний балл: {self.get_average():.2f} | "
                f"Тел: {self.__tel_number}")


student1 = Student("Иван", "ИС-21", [4, 5, 4], "+7-999-111-22-33")
student2 = Student("Анна", "ИС-22", [5, 5, 5], "+7-999-444-55-66")
student3 = Student("Петр", "ИС-21", [3, 4, 3], "+7-999-777-88-99")


student1.add_grade(5)
student3.add_grade(4)


print(student1.info())
print(student2.info())
print(student3.info())
```
