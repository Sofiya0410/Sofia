```python
# 1 lects
n = int(input("Введите количество человек: "))
if n <= 0:
    print("Ошибка")
else:
    scor= 0
    for i in range(1,n+1):
        age = int(input(f"Введите возраст {i} -го человека"))
        scor+=age
        print("Средний возраст:", scor/n)
```

```python
code = input("Введите цветовой код (например, a0351f): ").lstrip('#')

r = int(code[0:2], 16)
g = int(code[2:4], 16)
b = int(code[4:6], 16)

if r == g == b:
    if r == 0:
        print("черный")
    elif r == 255:
        print("белый")
    else:
        print("серый")
else:
    dr = (r - 255) ** 2 + g ** 2 + b ** 2
    dg = r ** 2 + (g - 255) ** 2 + b ** 2
    db = r ** 2 + g ** 2 + (b - 255) ** 2

    if dr < dg and dr < db:
        print("красный")
    elif dg < db:
        print("зеленый")
    else:
        print("синий")

```
