```python
# 3 задание
list = ["яблоко", "груша", "мандарин", "виноград","груша","банан","слива","абрикос","слива","яблоко"]

def count_occurrences(list):
   slovo = {}
   for i in list:
      if i in slovo:
          slovo[i] += 1
      else:
          slovo[i] = 1
   return slovo
result = count_occurrences(list)
print(result)
```
